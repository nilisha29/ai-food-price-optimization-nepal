# Nepal Food Commodity Data Exploration

This notebook documents the dataset used by the dynamic pricing system. It checks coverage, quality, commodity categories, and price trends before modeling.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

PROJECT_ROOT = Path.cwd().parent
DATA_PATH = PROJECT_ROOT / 'backend' / 'featured_dataset.csv'
df = pd.read_csv(DATA_PATH, parse_dates=['date'])
df.head()
```

```python
print(f'Rows: {len(df):,}')
print(f'Columns: {len(df.columns)}')
print(f'Date range: {df.date.min().date()} to {df.date.max().date()}')
print(f'Commodities: {df.commodity.nunique()}')
print(f'Markets: {df.market.nunique()}')
print(f'Missing values: {int(df.isna().sum().sum())}')
df.groupby(['category', 'commodity']).size().reset_index(name='rows').sort_values('rows', ascending=False)
```

```python
monthly = df.assign(month_period=df.date.dt.to_period('M')).groupby('month_period')['price_nrs'].mean()
monthly.plot(figsize=(12, 4), title='Average Nepal Food Commodity Price by Month')
plt.ylabel('Price (NRs/kg)')
plt.xlabel('Month')
plt.grid(alpha=0.2)
plt.show()
```
