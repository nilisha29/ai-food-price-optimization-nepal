# Pricing and Demand Model Evaluation

This notebook records the evaluation evidence used by the dashboard: error metrics, dataset sizes, and the most influential pricing features.

```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt

PROJECT_ROOT = Path.cwd().parent
BACKEND = PROJECT_ROOT / 'backend'
with open(BACKEND / 'model_metrics.json', encoding='utf-8') as file:
    price_metrics = json.load(file)
with open(BACKEND / 'demand_model_metrics.json', encoding='utf-8') as file:
    demand_metrics = json.load(file)
pd.DataFrame([price_metrics, demand_metrics], index=['price', 'demand'])[['accuracy', 'mae', 'rmse', 'r2', 'mape', 'train_rows', 'test_rows']]
```

```python
import joblib

model = joblib.load(BACKEND / 'xgboost_pricing_model.pkl')
features = price_metrics['features']
importance = (pd.DataFrame({'feature': features, 'importance': model.feature_importances_})
              .sort_values('importance', ascending=False).head(12))
importance
```

```python
importance.sort_values('importance').plot.barh(x='feature', y='importance', figsize=(9, 5), legend=False, title='Top Price Model Features')
plt.xlabel('XGBoost feature importance')
plt.tight_layout()
plt.show()
```
